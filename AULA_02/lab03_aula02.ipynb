```python
# ============================================================
# ATIVIDADE 3 - Heurística Gulosa x Solução Ótima
# Objetivo: comparar uma heurística simples com a solução ótima
# ============================================================

import itertools
import time
import random
import numpy as np


# ----------------------------------------------------------
# 1. FUNÇÃO PARA RESOLVER A MOCHILA POR FORÇA BRUTA
# ----------------------------------------------------------

def mochila_forca_bruta(pesos, valores, capacidade):
    """
    Testa todas as combinações possíveis e retorna o melhor valor.
    """
    n = len(pesos)

    melhor_valor = 0
    melhor_combinacao = None

    for combinacao in itertools.product([0, 1], repeat=n):

        peso_total = 0
        valor_total = 0

        for i in range(n):
            if combinacao[i] == 1:
                peso_total += pesos[i]
                valor_total += valores[i]

        if peso_total <= capacidade and valor_total > melhor_valor:
            melhor_valor = valor_total
            melhor_combinacao = combinacao

    return melhor_valor, melhor_combinacao


# ----------------------------------------------------------
# 2. HEURÍSTICA GULOSA
# ----------------------------------------------------------

def mochila_gulosa(pesos, valores, capacidade):
    """
    Resolve a mochila utilizando uma heurística gulosa.

    Os itens são ordenados pela razão valor/peso.
    """

    n = len(pesos)

    # Calcula valor/peso de cada item
    itens = []

    for i in range(n):
        if pesos[i] > 0:
            razao = valores[i] / pesos[i]
        else:
            razao = 0

        itens.append((razao, i))

    # Ordena da maior razão para a menor
    itens.sort(reverse=True)

    peso_total = 0
    valor_total = 0
    combinacao = [0] * n

    # Escolhe os itens enquanto houver espaço
    for razao, i in itens:

        if peso_total + pesos[i] <= capacidade:
            combinacao[i] = 1
            peso_total += pesos[i]
            valor_total += valores[i]

    return valor_total, tuple(combinacao)


# ----------------------------------------------------------
# 3. FUNÇÃO PARA CALCULAR O GAP
# ----------------------------------------------------------

def calcular_gap(valor_heuristica, valor_otimo):
    """
    Calcula o gap percentual entre a solução heurística
    e a solução ótima.
    """

    if valor_otimo == 0:
        return 0

    return ((valor_otimo - valor_heuristica) / valor_otimo) * 100


# ----------------------------------------------------------
# 4. GERANDO INSTÂNCIAS ALEATÓRIAS
# ----------------------------------------------------------

np.random.seed(42)

NUM_INSTANCIAS = 20
NUM_ITENS = 12
CAPACIDADE = 30

gaps = []

print("Rodando 20 instancias...")
print("=" * 75)


# ----------------------------------------------------------
# 5. EXECUTANDO AS 20 INSTÂNCIAS
# ----------------------------------------------------------

for k in range(NUM_INSTANCIAS):

    # Gera pesos aleatórios
    pesos = np.random.randint(1, 11, size=NUM_ITENS)

    # Gera valores aleatórios
    valores = np.random.randint(1, 21, size=NUM_ITENS)

    # ------------------------------------------------------
    # Solução ótima
    # ------------------------------------------------------

    inicio = time.time()

    otimo, combinacao_otima = mochila_forca_bruta(
        pesos,
        valores,
        CAPACIDADE
    )

    tempo_otimo = time.time() - inicio

    # ------------------------------------------------------
    # Solução heurística gulosa
    # ------------------------------------------------------

    inicio = time.time()

    heur, combinacao_heuristica = mochila_gulosa(
        pesos,
        valores,
        CAPACIDADE
    )

    tempo_heur = time.time() - inicio

    # ------------------------------------------------------
    # Calculando o GAP
    # ------------------------------------------------------

    gap = calcular_gap(heur, otimo)

    gaps.append(gap)

    # ------------------------------------------------------
    # Mostrando resultado da instância
    # ------------------------------------------------------

    print(
        f"Instancia {k+1:2d} | "
        f"Otimo: {otimo:4d} | "
        f"Gulosa: {heur:4d} | "
        f"Gap: {gap:5.1f}%"
    )


# ----------------------------------------------------------
# 6. ESTATÍSTICAS FINAIS
# ----------------------------------------------------------

print("=" * 75)

print(f"Gap medio: {np.mean(gaps):.2f}%")
print(f"Maior gap: {np.max(gaps):.2f}%")
print(f"Menor gap: {np.min(gaps):.2f}%")

print("=" * 75)


# O que entregar / discutir:
# 19.	Codigo completo (com a funcao calcular_gap implementada e o loop funcionando).
# 20.	Valor do gap medio obtido.
# 21.	Resposta: “A heuristica gulosa e boa o suficiente para este problema? Em quais situacoes voce usaria ela e em quais preferiria gastar mais tempo para achar o otimo?”
```
