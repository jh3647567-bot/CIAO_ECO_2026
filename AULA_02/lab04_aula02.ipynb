```python
import random

# ============================================================
# ATIVIDADE 4 - LISTA DE COMPRAS COM ORÇAMENTO LIMITADO
# ============================================================

# Produtos disponíveis
produtos = [
    "Arroz",
    "Feijão",
    "Leite",
    "Frango",
    "Frutas",
    "Café",
    "Sabonete",
    "Papel higiênico"
]

# Preço de cada produto
precos = [
    8,
    7,
    5,
    15,
    10,
    12,
    4,
    9
]

# Utilidade de cada produto
utilidades = [
    8,
    9,
    7,
    10,
    8,
    9,
    6,
    10
]

# Orçamento máximo
orcamento = 50

# Quantidade de produtos
n = len(produtos)


# ============================================================
# GERAÇÃO DE UMA SOLUÇÃO ALEATÓRIA
# ============================================================

solucao = [random.randint(0, 1) for _ in range(n)]

print("Solução aleatória:")
print(solucao)


# ============================================================
# CÁLCULO DO CUSTO E DA UTILIDADE
# ============================================================

custo_total = 0
utilidade_total = 0

for i in range(n):

    if solucao[i] == 1:
        custo_total += precos[i]
        utilidade_total += utilidades[i]


# ============================================================
# VERIFICAÇÃO DA RESTRIÇÃO
# ============================================================

factivel = custo_total <= orcamento


# ============================================================
# MOSTRANDO OS RESULTADOS
# ============================================================

print("\nProdutos escolhidos:")

for i in range(n):

    if solucao[i] == 1:
        print(
            "-",
            produtos[i],
            "| Preço: R$",
            precos[i],
            "| Utilidade:",
            utilidades[i]
        )

print("\nCusto total: R$", custo_total)
print("Utilidade total:", utilidade_total)
print("Orçamento disponível: R$", orcamento)


if factivel:

    print("Solução factível: SIM")
    print("Orçamento restante: R$", orcamento - custo_total)

else:

    print("Solução factível: NÃO")
    print("Orçamento excedido em: R$", custo_total - orcamento)
```

# Atividade 4: Modelagem de um Problema Real

## 1. Descrição do problema

Um problema comum do cotidiano é escolher quais produtos comprar quando temos um orçamento limitado.

Neste trabalho, vamos considerar uma situação em que uma pessoa possui R$ 50,00 para realizar suas compras. Existem vários produtos disponíveis, cada um com um preço e um nível de utilidade.

O objetivo é escolher quais produtos comprar de forma que a utilidade total seja a maior possível, sem ultrapassar o orçamento disponível.

Cada produto pode ser comprado uma única vez ou não ser comprado.

---

## 2. Modelagem formal

### 2.1 O que é uma solução?

Uma solução candidata é representada por uma sequência de números 0 e 1.

- 0 significa que o produto não será comprado.
- 1 significa que o produto será comprado.

Por exemplo:

[1, 0, 1, 0, 1, 0, 0, 1]

significa que os produtos 1, 3, 5 e 8 foram selecionados.

### 2.2 Espaço de busca

Neste problema existem 8 produtos.

Como cada produto possui duas possibilidades, comprar ou não comprar, existem:

2^8 = 256

combinações possíveis.

Portanto, o espaço de busca possui 256 soluções candidatas.

### 2.3 Função objetivo

O objetivo é maximizar a utilidade total dos produtos escolhidos.

A função objetivo pode ser representada por:

Maximizar Z = soma das utilidades dos produtos selecionados.

Quanto maior a utilidade total, melhor é a solução.

### 2.4 Restrições

A principal restrição é o orçamento.

O custo total dos produtos escolhidos não pode ultrapassar R$ 50,00.

Além disso, cada produto pode ser escolhido no máximo uma vez.

---

## 3. Classificação do problema

Este problema é semelhante ao Problema da Mochila, conhecido como Knapsack Problem.

Para uma quantidade pequena de produtos, podemos testar todas as combinações rapidamente. Porém, conforme aumentamos a quantidade de produtos, o número de possibilidades cresce exponencialmente.

Por exemplo:

- 8 produtos → 256 possibilidades;
- 20 produtos → 1.048.576 possibilidades;
- 30 produtos → 1.073.741.824 possibilidades.

Por esse motivo, a versão geral do problema pode ser considerada difícil e relacionada a problemas NP-difíceis.

A força bruta consegue encontrar a solução ótima, mas pode se tornar muito lenta quando o número de produtos aumenta.

---

## 4. Objetivo do código

O código Python irá gerar uma solução aleatória para a lista de compras.

Depois disso, o programa irá:

1. verificar quais produtos foram selecionados;
2. calcular o custo total;
3. calcular a utilidade total;
4. verificar se o orçamento foi respeitado;
5. informar se a solução é factível ou não.

## 5. Análise do resultado

O programa gerou uma solução aleatória para o problema de seleção dos produtos.

A solução foi representada por uma sequência de zeros e uns, em que 1 representa um produto selecionado e 0 representa um produto não selecionado.

Depois de gerar a solução, o programa calculou o custo total e a utilidade total dos produtos escolhidos.

Também foi verificado se o custo total respeitava o orçamento máximo de R$ 50,00.

Uma solução é considerada factível quando o custo total é menor ou igual ao orçamento disponível. Caso o custo ultrapasse R$ 50,00, a solução é considerada não factível.

A solução gerada aleatoriamente não necessariamente é a melhor solução possível. Para encontrar a melhor solução seria necessário comparar as diferentes combinações de produtos ou utilizar algum método de otimização.

## 6. Conclusão

A atividade mostrou como um problema do cotidiano pode ser transformado em um problema de otimização.

Neste caso, foi necessário escolher produtos tentando maximizar a utilidade sem ultrapassar um orçamento definido.

A representação por 0 e 1 facilita a construção das soluções e permite avaliar rapidamente se uma determinada combinação respeita as restrições.

O problema possui crescimento exponencial do espaço de busca, pois cada novo produto dobra o número de combinações possíveis. Por isso, embora seja simples resolver pequenas instâncias, problemas maiores podem exigir métodos de otimização mais eficientes.
